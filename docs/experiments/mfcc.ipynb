```python
import librosa
import librosa.display
import matplotlib.pyplot as plt

from google.colab import drive
```

```python
# Connect Google Drive

drive.mount('/content/drive')
```

```python
SONG_PATH = '/content/drive/MyDrive/songo/songo_datasets/deep_house_data/1-s1.aiff'
```

```python
song, sr = librosa.load(SONG_PATH)
```

```python
mfcc_song = librosa.feature.mfcc(song, n_mfcc=13, sr=sr)

plt.figure(figsize=(15, 8))
librosa.display.specshow(mfcc_song, x_axis="time", sr=sr)
plt.colorbar(format="%+2.f")
plt.show()
```
